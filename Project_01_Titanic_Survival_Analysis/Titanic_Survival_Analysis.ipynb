## Analysis & Insight

#### Question 1
How many passengers survived?

```python
df["Survived"].value_counts()
```

549 passengers did not survive, while 342 passengers survived. This indicates that the majority of passengers on the Titanic did not survive the disaster.

#### Question 2

Did gender affect survival?

```python
df.groupby("Sex")["Survived"].mean()
```

The analysis shows that female passengers had a much higher survival rate than male passengers. Approximately 74% of female passengers survived, compared to about 19% of male passengers. This suggests that gender was strongly associated with survival during the Titanic disaster.

#### Question 3
Did passenger class affect survival?

```python
df.groupby("Pclass")["Survived"].mean()
```

The analysis shows that First Class passengers had the highest survival rate (approximately 63%), followed by Second Class passengers (about 47%). Third Class passengers had the lowest survival rate (approximately 24%). This suggests that passenger class had a significant influence on survival during the Titanic disaster.

#### Question 4 
Did age affect survival?

```python
df.groupby("Survived")["Age"].mean()
```

The analysis indicates that passengers who survived were, on average, slightly younger than those who did not survive. Survivors had an average age of approximately 28 years, while non-survivors had an average age of about 31 years. This suggests that age may have had some influence on survival, although the difference is relatively small compared to factors such as passenger class and gender.

#### Question 5

Did the fare paid affect survival?

```python
df.groupby("Survived")["Fare"].mean()
```

The analysis indicates that passengers who survived paid higher fares on average than those who did not survive. Survivors paid an average fare of approximately 48, while non-survivors paid an average fare of about 22. This suggests that passengers who purchased more expensive tickets were more likely to survive the Titanic disaster.

#### Question 6

Did passengers traveling with siblings or spouses have a higher survival rate?

```python
df.groupby("SibSp")["Survived"].mean()
```

The analysis shows that survival rates varied across different SibSp groups. Passengers traveling with one sibling or spouse appeared to have a higher survival rate than some other groups, while those traveling in larger family groups had lower survival rates. This suggests that the number of siblings or spouses accompanying a passenger may have influenced survival, although the relationship is not consistent across all groups.

#### Question 7

Did passengers traveling with parents or children have a higher survival rate?

```python
df.groupby("Parch")["Survived"].mean()
```

The analysis indicates that passengers traveling with one, two, or three parents or children generally had higher survival rates than those traveling without parents or children. Passengers with three parents or children had the highest survival rate (60%), followed by those with one (55.1%) and two (50%). In contrast, passengers traveling without parents or children had a survival rate of approximately 34.4%, while those traveling with four or six parents or children did not survive. These findings suggest that traveling with a small family group may have been associated with a higher chance of survival, whereas very large family groups experienced lower survival rates.

##### Question 8

Which embarkation port had the highest survival rate?

```python
df.groupby("Embarked")["Survived"].mean()
```

The analysis indicates that the embarkation port was associated with different survival rates among passengers. Passengers who boarded the Titanic at Cherbourg (C) had the highest survival rate (approximately 55.4%), followed by Queenstown (Q) at about 39.0%. Southampton (S) had the lowest survival rate (approximately 33.7%). These findings suggest that the port of embarkation may have influenced survival outcomes, although the dataset does not provide enough information to explain the reasons for these differences.

#### Question 9

Did passengers with an assigned cabin have a higher survival rate?

```python
df.groupby(df["Cabin"].notnull())["Survived"].mean()
```

The analysis shows that passengers with assigned cabin have a higher survival rate with (approximately 66%) rate while the passengers without assigned cabin had the lower survival rate with (approximately 29%). These findings show that cabin assignment influence survival outcomes.

## Visualization 

#### 1. How did the number of survivors differ between female and male passengers?

```python
# Bar Chart 
import matplotlib.pyplot as plt

gender = ["Female", "Male"]
survivors = [233, 109]

plt.figure(figsize=(6, 5))

plt.bar(gender,survivors,color=["pink", "steelblue"])

plt.title("Titanic Survivors by Gender", fontsize=15, fontweight="bold")
plt.xlabel("Gender")
plt.ylabel("Number of Survivors")

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

#### 2. What ages were most common among the passengers?

```python
#Histogram
import matplotlib.pyplot as plt

#the layout, 6 inches wide and 5 inches tall\height
plt.figure(figsize=(6, 5))

#plot histogram with age column, seprate the age into 10 intervals, black color and transparency of 0.7
plt.hist(df["Age"], bins=10, edgecolor="black",alpha=0.7)

# label for title
plt.title("Age Distribution of Titanic Passengers")
#label for x 
plt.xlabel("Age")
#label for y
plt.ylabel("Frequency")

plt.grid(axis="y", linestyle="--", alpha=0.4)

#well space out layout
plt.tight_layout()
plt.show()
```
